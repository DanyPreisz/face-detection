# Detección de Rostros

Proyecto de visión por computadora listo para **Google Colab**.

**Pregunta:** ¿cómo detectar rostros en imágenes?

**Método:** OpenCV Haar Cascades o MediaPipe

**Stack:** opencv-python, matplotlib

Ejemplo con imagen de prueba.

## 0. Setup

```python
!pip install -q opencv-python-headless

import cv2
import matplotlib.pyplot as plt
import urllib.request
import numpy as np
print("OK")
```

## 1. Descargar imagen de ejemplo

```python
url = "https://upload.wikimedia.org/wikipedia/commons/thumb/5/50/Vd-Orig.jpg/800px-Vd-Orig.jpg"
urllib.request.urlretrieve(url, "ejemplo.jpg")
img = cv2.imread("ejemplo.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

## 2. Cargar clasificador Haar

```python
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
print("Clasificador cargado")
```

## 3. Detectar rostros

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5)
print(f"Rostros detectados: {len(faces)}")

for (x, y, w, h) in faces:
    cv2.rectangle(img_rgb, (x, y), (x+w, y+h), (0, 255, 0), 2)

plt.imshow(img_rgb)
plt.axis("off")
plt.title(f"{len(faces)} rostros")
plt.show()
```

## 4. Conclusiones

- OpenCV Haar Cascades es rápido y simple para detección de rostros.
- Ideal para empezar con visión por computadora.
- Se puede extender a MediaPipe o modelos deep learning (MTCNN, RetinaFace).
